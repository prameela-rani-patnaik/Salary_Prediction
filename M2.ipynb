```python
import pandas as pd
import numpy as np
```

```python
class M2:
    def __init__(self):
        self.m = 0
        self.b = 0

    def train(self, X_train, Y_train):
        num = 0
        den = 0

        x_bar = X_train.mean()
        y_bar = Y_train.mean()

        for i in range(len(X_train)):
            num += (X_train[i] - x_bar) * (Y_train[i] - y_bar)
            den += (X_train[i] - x_bar) ** 2

        self.m = num / den
        self.b = y_bar - self.m * x_bar

    def predict(self, X):
        return self.m * X + self.b
```
